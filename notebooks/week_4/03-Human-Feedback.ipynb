### Import Dependencies

```python
from langsmith import Client
```

### Attach Feedback to a specific trace in LangSmith

```python
client =  Client()
```

```python
RUN_ID = "019fb7fc-3b1c-7853-bbce-9f7a169f0cfb"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="This is a long form free text feedback left by Gaspar",
    feedback_source_type="api"
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment_with_score",
    score=1,
    value="This is a long form free text feedback left by Gaspar",
    feedback_source_type="api"
)
```
